```python
import numpy as np
import pandas as pd
import torch

@torch.no_grad()
def inspect_high_attention_weeks(
    batch, model,
    split="val",
    attn_quantile=0.8,
    n_patients=5,
):
    idx = batch.idx_val if split == "val" else batch.idx_tr

    S = batch.S[idx]
    C = batch.C[idx]
    y0p = batch.y0_phq[idx]
    y0p4 = batch.y0_p4[idx]
    y0lon = batch.y0_lon[idx]

    model.eval()
    _, _, _, attn, _ = model.forward_multi(S, C, y0p, y0p4, y0lon)
    # attn: (B,T)

    attn = attn.detach().cpu().numpy()
    T = attn.shape[1]

    rows = []
    for b in range(min(n_patients, attn.shape[0])):
        thr = np.quantile(attn[b], attn_quantile)
        for t in range(T):
            rows.append({
                "patient_idx": int(b),
                "week": int(t),
                "attn": float(attn[b, t]),
                "high_attn": bool(attn[b, t] >= thr),
            })

    df = pd.DataFrame(rows)
    return df
df_high = inspect_high_attention_weeks(
    batch, model,
    split="val",
    attn_quantile=0.8,
    n_patients=3,
)
display(df_high.head(30))
```

```python
# =========================================================
# PATCH: Robust logistic OR (handles singular matrix)
# =========================================================

import numpy as np
import pandas as pd
import os

from statsmodels.stats.multitest import multipletests

try:
    import statsmodels.api as sm
    _HAS_SM = True
except Exception as e:
    _HAS_SM = False
    sm = None
    print("[WARN] statsmodels import failed:", e)


def ensure_dir(path: str) -> str:
    os.makedirs(path, exist_ok=True)
    return path


def _drop_zero_var_cols(X: np.ndarray, cols: list[str], eps: float = 1e-12):
    std = X.std(axis=0)
    keep = std > eps
    X2 = X[:, keep]
    cols2 = [c for c, k in zip(cols, keep) if k]
    dropped = [c for c, k in zip(cols, keep) if not k]
    return X2, cols2, dropped


def _select_full_rank_subset(X: np.ndarray, cols: list[str], tol: float = 1e-10):
    """
    Greedy column selection: keep a column only if it increases matrix rank.
    p ~ 20~30이면 충분히 빠르게 돎.
    """
    keep_idx = []
    cur_rank = 0
    for j in range(X.shape[1]):
        cand_idx = keep_idx + [j]
        r = np.linalg.matrix_rank(X[:, cand_idx], tol=tol)
        if r > cur_rank:
            keep_idx.append(j)
            cur_rank = r

    X2 = X[:, keep_idx]
    cols2 = [cols[j] for j in keep_idx]
    dropped = [c for j, c in enumerate(cols) if j not in keep_idx]
    return X2, cols2, dropped, cur_rank


def run_service_logit_or(
    df: pd.DataFrame,
    outdir: str,
    prefix: str,
    baseline_mode: str = "all3",      # "single" | "all3"
    outcome_for_single: str | None = None,  # baseline_mode="single"일 때 필요
    drop_service_total_if_needed: bool = True,
    ridge_alpha: float = 1e-4,         # regularized fallback 강도
):
    """
    Robust OR:
    - 상수/제로분산 컬럼 제거
    - full-rank 컬럼 subset 선택
    - Logit fit이 singular면 GLM Binomial regularized로 fallback

    저장:
      service_or_{prefix}.csv
    """

    if not _HAS_SM:
        print("[SKIP] statsmodels not available")
        return None

    OUT = ensure_dir(outdir)

    service_cols = [c for c in df.columns if c.startswith("service")]
    if len(service_cols) == 0:
        raise ValueError("No service columns found in df")

    if "service_total" not in df.columns:
        raise ValueError("df must include service_total")
    if "target" not in df.columns:
        raise ValueError("df must include target")

    # baseline covariates
    if baseline_mode == "all3":
        base_cols = ["y0_PHQ", "y0_P4", "y0_Lon"]
    elif baseline_mode == "single":
        mapping = {"phq": "y0_PHQ", "p4": "y0_P4", "lon": "y0_Lon"}
        if outcome_for_single not in mapping:
            raise ValueError("outcome_for_single must be one of phq|p4|lon when baseline_mode='single'")
        base_cols = [mapping[outcome_for_single]]
    else:
        raise ValueError("baseline_mode must be 'single' or 'all3'")

    covar_cols = service_cols + base_cols + ["service_total"]

    y = df["target"].astype(int).values
    X = df[covar_cols].astype(float).values

    # ---- 1) 표준화 ----
    Xz = (X - X.mean(axis=0, keepdims=True)) / (X.std(axis=0, keepdims=True) + 1e-6)

    # ---- 2) 제로 분산 제거 ----
    Xz, cols_z, dropped_z = _drop_zero_var_cols(Xz, covar_cols)
    if dropped_z:
        print("[INFO] Dropped zero-variance cols:", dropped_z)

    # ---- 3) full-rank subset 선택 ----
    Xz, cols_r, dropped_r, rk = _select_full_rank_subset(Xz, cols_z)
    if dropped_r:
        print("[INFO] Dropped collinear cols (to make full-rank):", dropped_r)
    # intercept 추가
    X_sm = sm.add_constant(Xz, has_constant="add")

    # ---- 4) Logit 시도 ----
    used_cols = ["intercept"] + cols_r
    try:
        res = sm.Logit(y, X_sm).fit(disp=False)

        params = res.params
        bse    = res.bse
        pvals  = res.pvalues

        OR      = np.exp(params)
        OR_low  = np.exp(params - 1.96 * bse)
        OR_high = np.exp(params + 1.96 * bse)

        or_df = pd.DataFrame(
            {"beta": params, "OR": OR, "CI_low": OR_low, "CI_high": OR_high, "p_value": pvals},
            index=used_cols
        )

        # 서비스 항목 FDR
        svc_in_model = [c for c in cols_r if c.startswith("service")]
        if svc_in_model:
            svc_p = or_df.loc[svc_in_model, "p_value"].values
            _, qvals, _, _ = multipletests(svc_p, alpha=0.05, method="fdr_bh")
            or_df.loc[svc_in_model, "q_value"] = qvals

        csv_path = os.path.join(OUT, f"service_or_{prefix}.csv")
        or_df.to_csv(csv_path)
        print(f"[SAVE] OR table -> {csv_path}")
        return or_df

    except np.linalg.LinAlgError as e:
        print("[WARN] Logit failed with LinAlgError (singular). Falling back to GLM regularized.")
        print("       Error:", e)

    # ---- 5) fallback: GLM Binomial regularized ----
    # regularized는 CI/pvalue가 기본적으로 깔끔히 안 나와서 OR만 저장(대신 안 터짐)
    glm = sm.GLM(y, X_sm, family=sm.families.Binomial())
    try:
        res2 = glm.fit_regularized(alpha=ridge_alpha, L1_wt=0.0)  # ridge
        params = res2.params
        OR = np.exp(params)

        or_df = pd.DataFrame(
            {"beta": params, "OR": OR, "CI_low": np.nan, "CI_high": np.nan, "p_value": np.nan},
            index=used_cols
        )

        csv_path = os.path.join(OUT, f"service_or_{prefix}.csv")
        or_df.to_csv(csv_path)
        print(f"[SAVE] OR table (REGULARIZED; CI/p=N/A) -> {csv_path}")
        return or_df

    except Exception as e:
        print("[FAIL] GLM regularized fallback also failed:", e)
        return None
```

```python
import os
import numpy as np
import pandas as pd
import torch

@torch.no_grad()
def export_ch_attn_overall_values(batch, model, outdir, split="val", fname=None):
    """
    plot_ch_attn_overall_{split}.png의 기반 숫자(mean over B,T)를
    표 형태로 저장/출력.
    """
    os.makedirs(outdir, exist_ok=True)
    idx = batch.idx_val if split == "val" else batch.idx_tr

    S = batch.S[idx]
    C = batch.C[idx]
    y0p = batch.y0_phq[idx]
    y0p4 = batch.y0_p4[idx]
    y0lon = batch.y0_lon[idx]

    model.eval()
    _, _, _, _, ch_alpha = model.forward_multi(S, C, y0p, y0p4, y0lon)  # (B,T,6)

    ch = ch_alpha.detach().cpu().numpy()
    mean_alpha = ch.mean(axis=(0, 1))  # (6,)

    check_names = batch.stats.get("check_basenames", batch.stats.get("check_cols"))
    df = pd.DataFrame({
        "channel": check_names,
        "mean_ch_attn_BT": mean_alpha
    })

    if fname is None:
        fname = f"ch_attn_overall_{split}_values.csv"
    out_path = os.path.join(outdir, fname)
    df.to_csv(out_path, index=False)

    print("[SAVE]", out_path)
    return df
df_ch_overall = approach3.export_ch_attn_overall_values(batch, model, OUT, split="val")
display()
```

```python
def _p(path):  # 짧게 쓰려고
    return os.path.join(OUT, path)

def pretty_print_importance_tables(OUT, exp_tag, topk=12):
    imp_dir = os.path.join(OUT, "phq_importance")

    overall_path  = os.path.join(imp_dir, f"overall_{exp_tag}.csv")
    percls_path   = os.path.join(imp_dir, f"per_class_top_{exp_tag}.csv")
    contrast_path = os.path.join(imp_dir, f"contrast_{exp_tag}.csv")

    df_overall  = pd.read_csv(overall_path)
    df_percls   = pd.read_csv(percls_path)
    df_contrast = pd.read_csv(contrast_path)

    print("\n" + "="*90)
    print(f"[PHQ Importance Tables]  exp_tag={exp_tag}")
    print("="*90)

    # 1) Overall Top-k
    print(f"\n(1) Overall Top-{topk} (mean |coef| across classes; scaled이면 scaled 기준)")
    print(df_overall.head(topk).to_string(index=False))

    # 2) Per-class Top-k
    print(f"\n(2) Per-class Top-{topk}")
    for cls in df_percls["class"].unique():
        print(f"\n  - class = {cls}")
        sub = df_percls[df_percls["class"] == cls].head(topk)
        # scaled/coef 둘 다 있을 수 있어서 있는 컬럼만 예쁘게
        cols = [c for c in ["feature","coef","abs_coef","scaled_coef","abs_scaled_coef"] if c in sub.columns]
        print(sub[cols].to_string(index=False))

    # 3) Contrast (worse - improved)
    print(f"\n(3) Contrast Top-{topk} (coef_worse_minus_improved)")
    cols = [c for c in ["feature","coef_worse_minus_improved","abs"] if c in df_contrast.columns]
    print(df_contrast[cols].head(topk).to_string(index=False))

    return df_overall, df_percls, df_contrast


def make_one_clean_table(OUT, exp_tag, topk=12, save_name=None):
    """
    보고서 3개를 한 장짜리 '요약 표'로 합쳐서 보여주고 저장까지.
    - overall topk
    - per-class topk(각 class에서 topk)
    - contrast topk
    """
    imp_dir = os.path.join(OUT, "phq_importance")
    df_overall  = pd.read_csv(os.path.join(imp_dir, f"overall_{exp_tag}.csv")).head(topk).copy()
    df_percls   = pd.read_csv(os.path.join(imp_dir, f"per_class_top_{exp_tag}.csv")).copy()
    df_contrast = pd.read_csv(os.path.join(imp_dir, f"contrast_{exp_tag}.csv")).head(topk).copy()

    # overall
    df_overall["section"] = f"OVERALL_TOP{topk}"
    df_overall = df_overall[["section", "feature", "importance"]]

    # per-class (class별 topk만)
    out = []
    for cls in df_percls["class"].unique():
        sub = df_percls[df_percls["class"] == cls].head(topk).copy()
        sub["section"] = f"CLASS_{cls}_TOP{topk}"
        # coef/abs/scale 중 있는거만
        cols = ["section","class","feature"]
        for c in ["coef","abs_coef","scaled_coef","abs_scaled_coef"]:
            if c in sub.columns:
                cols.append(c)
        out.append(sub[cols])
    df_percls_small = pd.concat(out, axis=0, ignore_index=True)

    # contrast
    df_contrast["section"] = f"CONTRAST(worse-improved)_TOP{topk}"
    df_contrast = df_contrast[["section","feature","coef_worse_minus_improved","abs"]]

    # 합치기 (컬럼 달라도 concat 가능하게)
    df_all = pd.concat([df_overall, df_percls_small, df_contrast], axis=0, ignore_index=True)

    print("\n" + "="*90)
    print(f"[ONE CLEAN TABLE] exp_tag={exp_tag}")
    print("="*90)
    print(df_all.to_string(index=False))

    if save_name is None:
        save_name = f"PHQ_IMPORTANCE_CLEAN_{exp_tag}_TOP{topk}.csv"
    df_all.to_csv(os.path.join(imp_dir, save_name), index=False, encoding="utf-8")
    print(f"\nSaved: {os.path.join(imp_dir, save_name)}")

    return df_all
```

```python
from math import log

def binary_entropy(p):
    p = np.clip(p, 1e-8, 1-1e-8)
    return -(p*log(p) + (1-p)*log(1-p))

def channel_report(Craw, dY, check_names):
    # Craw: (N,T,C)
    N,T,C = Craw.shape
    rep = []

    for c in range(C):
        x = Craw[:,:,c]  # (N,T)

        # 전체 분산 + 환자내/환자간 분해
        mean_i = x.mean(axis=1)                 # (N,)
        var_within = ((x - mean_i[:,None])**2).mean()   # within
        var_between = mean_i.var()              # between
        var_total = x.var()

        # 변화량(transition) 빈도: 이진일 때 특히
        # (값이 이진이 아닐 수도 있으니 "변화 여부"로 계산)
        diff = (x[:,1:] != x[:,:-1]).astype(float)
        flip_rate = diff.mean()  # 전체 주차에서 값이 바뀌는 비율

        # 이진이면 엔트로피(주 단위 평균)
        uniq = np.unique(x)
        ent = np.nan
        if set(uniq.tolist()).issubset(set([0,1])):
            p = x.mean()
            ent = binary_entropy(p)

        rep.append(dict(
            channel=check_names[c] if check_names else f"check{c+1}",
            var_total=float(var_total),
            var_between=float(var_between),
            var_within=float(var_within),
            flip_rate=float(flip_rate),
            binary_entropy=float(ent) if ent==ent else np.nan,
            uniq_values=str(sorted(uniq.tolist())[:10])
        ))

    return pd.DataFrame(rep).sort_values("var_total", ascending=False)

df_ch = channel_report(val["Craw"], val["dY"], check_names)
df_ch
```

```python
def time_alignment_report(Craw, attn, dY, check_names, target_channel="check3", topk=5):
    # Craw:(N,T,C), attn:(N,T), dY:(N,)
    N,T,C = Craw.shape
    ch = check_names.index(target_channel) if isinstance(target_channel,str) else int(target_channel)

    x = Craw[:,:,ch]  # (N,T)
    # check3가 이진이라고 가정하고 "event week" = flip week로 정의
    flip = (x[:,1:] != x[:,:-1]).astype(int)          # (N,T-1)
    flip = np.concatenate([np.zeros((N,1),dtype=int), flip], axis=1)  # (N,T) week t에서 flip 발생 여부

    rows = []
    for cls in [0,1,2]:
        m = (dY==cls)
        if m.sum()==0: 
            continue

        attn_m = attn[m]        # (n,T)
        x_m    = x[m]
        flip_m = flip[m]

        # top-k attention week가 check3=1인지 / flip인지 비율
        top_idx = np.argsort(attn_m, axis=1)[:,-topk:]  # (n,k)
        # gather
        v_top = np.take_along_axis(x_m, top_idx, axis=1)
        f_top = np.take_along_axis(flip_m, top_idx, axis=1)

        rows.append(dict(
            group=["improved","same","worse"][cls],
            n=int(m.sum()),
            mean_attn_entropy=float((-attn_m*np.log(np.clip(attn_m,1e-8,1))).sum(axis=1).mean()),
            topk_mean_check3=float(v_top.mean()),            # topk 주에서 check3 평균 (이진이면 "1 비율")
            topk_flip_rate=float((f_top>0).mean()),          # topk 주에서 flip 발생 비율
            overall_check3_mean=float(x_m.mean()),
            overall_flip_rate=float(flip_m.mean()),
        ))

    return pd.DataFrame(rows)

align_df = time_alignment_report(val["Craw"], val["attn"], val["dY"], check_names, target_channel="check3", topk=5)
align_df
```

```python
def variance_decomposition(Craw, check_names):
    """
    Craw: (N,T,C)
    """
    N,T,C = Craw.shape
    rows = []

    for c in range(C):
        x = Craw[:,:,c]               # (N,T)
        mean_i = x.mean(axis=1)       # 환자별 평균

        var_between = mean_i.var()    # 환자 간 분산
        var_within  = ((x - mean_i[:,None])**2).mean()  # 환자 내 분산
        ratio = var_between / (var_within + 1e-8)

        rows.append(dict(
            channel=check_names[c],
            var_between=var_between,
            var_within=var_within,
            between_over_within=ratio
        ))

    return pd.DataFrame(rows).sort_values("between_over_within", ascending=False)

variance_decomposition(val["Craw"], check_names)
```

```python
def dynamics_report(Craw, check_names):
    """
    변화량/flip 기반 동학 비교
    """
    N,T,C = Craw.shape
    rows = []

    for c in range(C):
        x = Craw[:,:,c]
        # 값이 바뀌는 비율
        flip = (x[:,1:] != x[:,:-1]).astype(float)
        flip_rate = flip.mean()

        # 변화 크기 (연속형도 대응)
        delta = np.abs(x[:,1:] - x[:,:-1])
        mean_delta = delta.mean()

        rows.append(dict(
            channel=check_names[c],
            flip_rate=flip_rate,
            mean_abs_delta=mean_delta
        ))

    return pd.DataFrame(rows).sort_values("flip_rate", ascending=False)

dynamics_report(val["Craw"], check_names)
```

```python
def channel_correlation(Craw, check_names):
    """
    채널 평균값 기준 상관 (환자 레벨)
    """
    N,T,C = Craw.shape
    X = Craw.mean(axis=1)  # (N,C)

    corr = np.corrcoef(X.T)
    corr_df = pd.DataFrame(corr, index=check_names, columns=check_names)
    return corr_df

corr_df = channel_correlation(val["Craw"], check_names)
corr_df
```
