# **analysis_util.py**

````python
# =========================
# analysis_util 실행/표시용 (ipynb)
# - "값을 바꾸는 실험" (counterfactual/ICE/FD/permutation) 은 실행 안 함
# - 공병/상호작용/성능/캘리브레이션/서브그룹/어텐션 플롯만 실행 + 노트북에 표시
# =========================

import os, json
import pandas as pd
from IPython.display import display, Image, Markdown

OUTA = OUT  
os.makedirs(OUTA, exist_ok=True)

def notebook_run_analysis_util_no_perturb(raw2_csv, batch, model, outdir=OUTA, split="val",
                                          run_interactions=True,
                                          run_basic=True,
                                          run_reliability=True,
                                          run_subgroup=True,
                                          run_attn=True):
    """
    perturbation 없이(=서비스 값 바꾸는 실험 없이) 돌아가는 것만 실행.
    """

    display(Markdown("## (analysis_util) Running non-perturbation analyses"))

    # 1) 상호작용(공병 + Δ cross-tab + prob corr)
    if run_interactions:
        try:
            run_full_interaction_analysis(raw2_csv, batch, model, outdir)
        except Exception as e:
            print("[WARN] interaction analysis failed:", e)

    # 2) 기본 성능 리포트 + CM + misclassified.json
    if run_basic:
        try:
            basic_report(model, batch, outdir=outdir)
        except Exception as e:
            print("[WARN] basic_report failed:", e)

    # 3) 캘리브레이션(reliability + ECE)
    if run_reliability:
        try:
            reliability_diagram(model, batch, outdir=outdir, n_bins=15)
        except Exception as e:
            print("[WARN] reliability_diagram failed:", e)

    # 4) 서브그룹(y0 bin별 acc/f1)
    if run_subgroup:
        try:
            subgroup_by_y0(model, batch, outdir=outdir)
        except Exception as e:
            print("[WARN] subgroup_by_y0 failed:", e)

    # 5) 어텐션 플롯
    if run_attn:
        try:
            plot_attn_by_delta_phq(batch, model, outdir=outdir, split=split)
            plot_attn_by_y0_phq(batch, model, outdir=outdir, split=split)
        except Exception as e:
            print("[WARN] attention plots failed:", e)

    display(Markdown(f"✅ Saved outputs under: `{outdir}`"))

    # =========================
    # ipynb에 결과 띄우기
    # =========================
    display(Markdown("## (analysis_util) Key plots"))

    plot_files = [
        f"attn_by_dPHQ_{split}.png",
        f"attn_by_y0PHQ_{split}.png",
        "cm_val.png",
        "reliability.png",
    ]
    for fn in plot_files:
        p = os.path.join(outdir, fn)
        if os.path.exists(p):
            display(Markdown(f"**{fn}**"))
            display(Image(filename=p))
        else:
            print("[MISS]", fn)

    display(Markdown("## (analysis_util) Key tables / json previews"))

    csv_files = [
        "xtab_baseline_PHQ_P4.csv",
        "xtab_baseline_PHQ_Loneliness.csv",
        "xtab_true_PHQ_P4.csv",
        "xtab_true_PHQ_Loneliness.csv",
        "xtab_pred_PHQ_P4.csv",
        "xtab_pred_PHQ_Loneliness.csv",
    ]
    for fn in csv_files:
        p = os.path.join(outdir, fn)
        if os.path.exists(p):
            display(Markdown(f"**{fn}**"))
            display(pd.read_csv(p).head())
        else:
            print("[MISS]", fn)

    # interactions_delta.json (prob_corr / conditional 일부)
    p_json = os.path.join(outdir, "interactions_delta.json")
    if os.path.exists(p_json):
        display(Markdown("### interactions_delta.json (prob_corr)"))
        with open(p_json, "r", encoding="utf-8") as f:
            js = json.load(f)
        display(js.get("prob_corr", {}))

        display(Markdown("### interactions_delta.json (conditional: head)"))
        cond = js.get("conditional", {})

        display({k: cond[k] for k in list(cond.keys())[:2]})

    # misclassified.json
    p_mis = os.path.join(outdir, "misclassified.json")
    if os.path.exists(p_mis):
        display(Markdown("### misclassified.json (top few)"))
        with open(p_mis, "r", encoding="utf-8") as f:
            mis = json.load(f)

        preview = {k: mis[k] for k in ["count_val","acc","f1_macro","wrong_top20_indices","wrong_top20_pred","wrong_top20_true","wrong_top20_conf"] if k in mis}
        display(preview)

    # subgroup json
    p_sub = os.path.join(outdir, "subgroup_y0.json")
    if os.path.exists(p_sub):
        display(Markdown("### subgroup_y0.json"))
        with open(p_sub, "r", encoding="utf-8") as f:
            sub = json.load(f)
        display(pd.DataFrame(sub.get("rows", [])))

# 실행 
notebook_run_analysis_util_no_perturb(
    raw2_csv=RAW2,
    batch=batch,
    model=model,
    outdir=OUTA,
    split="val",
    run_interactions=True,
    run_basic=True,
    run_reliability=True,
    run_subgroup=True,
    run_attn=True,
)
````

````python
# =========================
# attn_check3_full_pipeline 실행/표시용 (ipynb)
# - per_patient_features 생성 + improved vs else 통계 + validation suite(V1/V3/V4)
# - 결과 파일 저장 + 노트북에 표/요약 출력
# =========================

import os, json, importlib
import numpy as np
import pandas as pd
from IPython.display import display, Markdown

import attn_check3_full_pipeline as fp
importlib.reload(fp)

def notebook_run_attn_check3_full_pipeline(
    batch,
    model,
    outdir,
    split="val",
    target_channel="check3",
    improved_label=0,

    # group tests
    smooth_feat=1,
    n_boot=2000,
    n_perm=5000,
    seed=0,

    # validation
    smooth_attn=3,
    n_boot_auc=3000,
    half_width=1,
    n_random=50,
    n_perm_window=8000,
    window_mode="set_patient_mean",
    ablation_modes=None,

    # display
    show_tables=True,
    n_head=10,
):
    os.makedirs(outdir, exist_ok=True)

    display(Markdown("## (attn_check3_full_pipeline) Running full pipeline"))

    group_cfg = fp.GroupTestConfig(
        split=split,
        target_channel=target_channel,
        smooth=smooth_feat,
        n_boot=n_boot,
        n_perm=n_perm,
        seed=seed,
        improved_label=improved_label,
    )
    val_cfg = fp.ValidationConfig(
        split=split,
        target_channel=target_channel,
        improved_label=improved_label,
        smooth_attn=smooth_attn,
        n_boot_auc=n_boot_auc,
        half_width=half_width,
        n_random=n_random,
        n_perm_window=n_perm_window,
        window_mode=window_mode,
        ablation_modes=ablation_modes,
        seed=seed,
    )
    cfg = fp.FullPipelineConfig(group=group_cfg, val=val_cfg)

    out = fp.run_all(batch=batch, model=model, outdir=outdir, cfg=cfg)

    paths = out.get("paths", {})
    display(Markdown(f"✅ Saved outputs under: `{outdir}`"))

    display(Markdown("## Output files"))
    for k, p in paths.items():
        print(f"[{'OK' if os.path.exists(p) else 'MISS'}] {k}: {p}")

    if show_tables:
        display(Markdown("## Key tables (head)"))

        if "df_features" in out:
            display(Markdown("### per_patient_features (df_features)"))
            display(out["df_features"].head(n_head))

        if "df_tests" in out:
            display(Markdown("### stats_tests (df_tests)"))
            display(out["df_tests"].head(n_head))

        cand_path = paths.get("candidates_csv", "")
        if cand_path and os.path.exists(cand_path):
            display(Markdown("### candidate_findings"))
            display(pd.read_csv(cand_path).head(n_head))
        else:
            print("[MISS] candidates_csv")

        marker_path = paths.get("marker_csv", "")
        if marker_path and os.path.exists(marker_path):
            display(Markdown("### marker_ch3_dpeak"))
            display(pd.read_csv(marker_path).head(n_head))
        else:
            print("[MISS] marker_csv")

        ab_path = paths.get("ablation_csv", "")
        if ab_path and os.path.exists(ab_path):
            display(Markdown("### model_ablation"))
            display(pd.read_csv(ab_path).head(n_head))
        else:
            print("[MISS] ablation_csv")

        win_path = paths.get("window_csv", "")
        if win_path and os.path.exists(win_path):
            display(Markdown("### window_faithfulness"))
            display(pd.read_csv(win_path).head(n_head))
        else:
            print("[MISS] window_csv")

    display(Markdown("## Validation summary (printed)"))
    v = out.get("validation", {})
    v1 = v.get("v1_auc", {})
    v4 = v.get("v4_window_faithfulness", {})

    if v1:
        display(Markdown("### V1 Univariate AUC (marker_ch3_dpeak_val)"))
        display(pd.DataFrame([v1]))
    else:
        print("[MISS] v1_auc")

    if v4:
        display(Markdown("### V4 Window faithfulness (targeted vs random occlusion)"))
        display(pd.DataFrame([v4]))
    else:
        print("[MISS] v4_window_faithfulness")

    if "v3_model_ablation" in v:
        display(Markdown("### V3 Model-level ablation on INPUT check3"))
        try:
            display(v["v3_model_ablation"])
        except Exception:
            print(v["v3_model_ablation"])
    else:
        print("[MISS] v3_model_ablation")

    return out


# ===== 실행 예시 =====

OUT_FULL = "figs_attn_check3_full_pipeline"
result = notebook_run_attn_check3_full_pipeline(
    batch=batch,
    model=model,
    outdir=OUT_FULL,
    split="val",
    target_channel="check3",
    improved_label=0,

    smooth_feat=1,
    n_boot=2000,
    n_perm=5000,
    seed=42,

    smooth_attn=3,
    n_boot_auc=3000,
    half_width=1,
    n_random=50,
    n_perm_window=8000,
    window_mode="set_patient_mean",
    ablation_modes=["permute_time", "set_patient_mean", "set_global_mode"],

    show_tables=True,
    n_head=10,
)
````

# **service_postevent_end2end**

````python
# =========================
# service_postevent_end2end 실행/표시용 (ipynb)
# - from_batch: batch.pt -> merged 3종 생성 -> screening + multivar logit -> 결과 표시
# - from_merged: merged 3종이 이미 있으면 분석만 실행
# =========================

import os, importlib, json
import pandas as pd
from IPython.display import display, Markdown

import service_postevent_end2end as spe
importlib.reload(spe)

def _show_csv_head(path: str, n_head: int = 20, title: str = ""):
    if os.path.exists(path):
        if title:
            display(Markdown(f"### {title}"))
        display(Markdown(f"`{path}`"))
        display(pd.read_csv(path).head(n_head))
    else:
        print("[MISS]", path)

def notebook_run_service_postevent_from_batch(
    BATCH_PATH: str,
    POSTEVENT_CSV: str,
    OUTDIR: str,
    SPLIT: str = "val",
    T_CUT: int = 34,
    RECENT_WINDOW: int = 3,
    Y_COL: str = "post_event",
    P_SCREEN: float = 0.10,
    COUNT_TRANSFORM: str = "log1p",
    show_tables: bool = True,
    n_head: int = 20,
):
    os.makedirs(OUTDIR, exist_ok=True)

    if spe.torch is None:
        raise RuntimeError("torch import failed inside service_postevent_end2end.py")

    display(Markdown("## (service_postevent_end2end) from_batch"))

    # 1) batch load
    batch = spe.torch.load(BATCH_PATH, map_location="cpu")
    service_cols = list(batch.stats["service_cols"])

    # 2) merged csvs 생성
    merged_paths = spe.make_merged_tables_from_batch(
        batch=batch,
        postevent_csv=POSTEVENT_CSV,
        outdir=OUTDIR,
        split=SPLIT,
        t_cut=T_CUT,
        recent_window=RECENT_WINDOW,
        post_event_col=Y_COL,
    )

    # 3) 분석 실행 (screening + multivar logit)
    summary_path = spe.run_analysis_from_merged(
        binary_csv=merged_paths["binary"],
        count_csv=merged_paths["count"],
        recent_csv=merged_paths["recent"],
        outdir=OUTDIR,
        service_cols=service_cols,
        y_col=Y_COL,
        p_screen=P_SCREEN,
        count_transform=COUNT_TRANSFORM,
    )

    display(Markdown(f"✅ Saved outputs under: `{OUTDIR}`"))
    print("[OK] run_summary:", summary_path)

    if show_tables:
        _show_csv_head(summary_path, n_head=n_head, title="run_summary")

        _show_csv_head(os.path.join(OUTDIR, "screening_binary.csv"), n_head=n_head, title="screening_binary")
        _show_csv_head(os.path.join(OUTDIR, "logit_binary.csv"), n_head=n_head, title="logit_binary")

        _show_csv_head(os.path.join(OUTDIR, "screening_count.csv"), n_head=n_head, title="screening_count")
        _show_csv_head(os.path.join(OUTDIR, "logit_count.csv"), n_head=n_head, title="logit_count")

        _show_csv_head(os.path.join(OUTDIR, "screening_recent.csv"), n_head=n_head, title="screening_recent")
        _show_csv_head(os.path.join(OUTDIR, "logit_recent.csv"), n_head=n_head, title="logit_recent")

    return {
        "outdir": OUTDIR,
        "merged_binary": merged_paths["binary"],
        "merged_count": merged_paths["count"],
        "merged_recent": merged_paths["recent"],
        "run_summary": summary_path,
        "service_cols": service_cols,
    }


def notebook_run_service_postevent_from_merged(
    BINARY_CSV: str,
    COUNT_CSV: str,
    RECENT_CSV: str,
    OUTDIR: str,
    SERVICE_COLS: list,
    Y_COL: str = "post_event",
    P_SCREEN: float = 0.10,
    COUNT_TRANSFORM: str = "log1p",
    show_tables: bool = True,
    n_head: int = 20,
):
    os.makedirs(OUTDIR, exist_ok=True)

    display(Markdown("## (service_postevent_end2end) from_merged"))

    summary_path = spe.run_analysis_from_merged(
        binary_csv=BINARY_CSV,
        count_csv=COUNT_CSV,
        recent_csv=RECENT_CSV,
        outdir=OUTDIR,
        service_cols=SERVICE_COLS,
        y_col=Y_COL,
        p_screen=P_SCREEN,
        count_transform=COUNT_TRANSFORM,
    )

    display(Markdown(f"✅ Saved outputs under: `{OUTDIR}`"))
    print("[OK] run_summary:", summary_path)

    if show_tables:
        _show_csv_head(summary_path, n_head=n_head, title="run_summary")

        _show_csv_head(os.path.join(OUTDIR, "screening_binary.csv"), n_head=n_head, title="screening_binary")
        _show_csv_head(os.path.join(OUTDIR, "logit_binary.csv"), n_head=n_head, title="logit_binary")

        _show_csv_head(os.path.join(OUTDIR, "screening_count.csv"), n_head=n_head, title="screening_count")
        _show_csv_head(os.path.join(OUTDIR, "logit_count.csv"), n_head=n_head, title="logit_count")

        _show_csv_head(os.path.join(OUTDIR, "screening_recent.csv"), n_head=n_head, title="screening_recent")
        _show_csv_head(os.path.join(OUTDIR, "logit_recent.csv"), n_head=n_head, title="logit_recent")

    return {"outdir": OUTDIR, "run_summary": summary_path}


# =========================
# 실행 예시 (from_batch)
# =========================
BATCH_PATH = "figs_phq_multi_attn/batch.pt"
POSTEVENT_CSV = "figs_check3_peakwindows/per_patient_windows_val.csv"

OUT_SVC = "figs_check3_peakwindows/service_postevent_end2end"

out_svc = notebook_run_service_postevent_from_batch(
    BATCH_PATH=BATCH_PATH,  # 예: "/home/nakyung/projects/DILFINAL/batch.pt"
    POSTEVENT_CSV=os.path.join("figs_check3_peakwindows", "per_patient_windows_val.csv"),
    OUTDIR=OUT_SVC,
    SPLIT="val",
    T_CUT=34,
    RECENT_WINDOW=3,
    Y_COL="post_event",
    P_SCREEN=0.10,
    COUNT_TRANSFORM="log1p",
    show_tables=True,
    n_head=20,
)
````
